```python
import xmlrpc.client
```

```python
HOST = "127.0.0.1"
PORT = 6001
```

```python
def main():
    proxy = xmlrpc.client.ServerProxy(f"http://{HOST}:{PORT}/")

    print("Calling remote add(4,7)->",proxy.add(4,7))
    print("calling remote factorial->",proxy.fac(7))
    print("calling remote reverse_string->",proxy.reverse_string("fwgeger"))
```

```python
if __name__ == "__main__":
    main()
```
