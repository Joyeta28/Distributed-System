```python
from xmlrpc.server import SimpleXMLRPCServer
```

```python
HOST = "127.0.0.1"
PORT = 6001
```

```python
def add(a,b):
    print(f"[SERVER] add({a},{b}) called remotely")
    return a+b
```

```python
def fac(n):
    print(f"[SERVER] fac({n}) was called remotely")
    result = 1
    for i in range(1,n+1):
        result*=i
    return result
```

```python
def reverse_string(s):
    print(f"[SERVER] revere_string({s}) was called remotely")
    return s[::-1]
```

```python
def main():
    server = SimpleXMLRPCServer((HOST,PORT),allow_none=True)
    server.register_function(add,"add")
    server.register_function(fac,"fac")
    server.register_function(reverse_string,"reverse_string")
    print(f"[SERVER] RPC server listening on {HOST}:{PORT}")
    print("[SERVER] Exposed procedures: add, factorial, reverse_string")
    server.serve_forever()
```

```python
if __name__ == "__main__":
    main()
```
