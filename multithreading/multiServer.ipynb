```python
import socket
import threading
import time
```

```python
HOST = "127.0.0.1"
PORT = 60001
```

```python
def handle_client(conn,addr):
    with conn:
        print(f"Worker-{threading.get_ident} handling address {addr}")
        request = conn.recv(1024).decode()
        time.sleep(2)
        
        reply = f"Processed {request} by worker-{threading.get_ident}"
        conn.sendall(reply.encode())
        print(f"Worker-{threading.get_ident} have processed {addr}")
```

```python
def main():
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as server_socket:
        server_socket.setsockopt(socket.SOL_SOCKET,socket.SO_REUSEADDR,1)
        server_socket.bind((HOST,PORT))
        server_socket.listen(5)
        print(f"[Dispatcher] is listening on {HOST}:{PORT}")
        while True:
            conn,addr = server_socket.accept()
            print(f"[Dispatcher] Accepted {addr},swaning worker thread")
            worker = threading.Thread(target = handle_client, args=(conn,addr))
            worker.start()
```

```python
if __name__ == "__main__":
    main()
```
