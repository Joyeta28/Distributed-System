```python
import socket
import threading
import time
```

```python
HOST = "127.0.0.1"
PORT = 65001
```

```python
n = 5
```

```python
def worker(i):
    with socket.socket(socket.AF_INET,socket.SOCK_STREAM) as client_socket:
        client_socket.connect((HOST,PORT))
        client_socket.sendall(f"request{i}".encode())
        reply = client_socket.recv(1024).decode()
        print(f"[CLIENT {i}] {reply}")
```

```python
def main():
    start = time.time()
    thread = [threading.Thread(target = worker,args=(i,)) for i in range(n)]
    for t in thread:
        t.start()
    for t in thread:
        t.join()
    elapsed = time.time()-start
    print(f"\nAll requests completed in {elapsed:.2f} seconds")
```

```python
if __name__ == "__main__":
    main()
    
```
